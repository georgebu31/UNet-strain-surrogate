# Exciton drift-diffusion solver demo
Runs the exciton density solver on the strain field from the FDM demo, using
a Gaussian excitation beam as the generation source, and plots the resulting
exciton density map.

```python
import numpy as np
import matplotlib.pyplot as plt

from fdm_solver import get_w_strain, draw_circle
from exciton_solver import solve_exciton_diffusion
```

## Calculate the strain field `demo_fdm_solver`

```python
NPZ_FILENAME   = "swg_256.npz"
NPZ_ARRAY_NAME = 'array'
width  = 12e-6
height = 12e-6

support_data = np.load(NPZ_FILENAME)
if NPZ_ARRAY_NAME in support_data:
    support_map = support_data[NPZ_ARRAY_NAME]
else:
    support_map = support_data[support_data.files[0]]

support_map = np.where(support_map > 0.5, 1, 0)

strain, X, Y = get_w_strain(support_map, width)
```

## Solve for the steady-state exciton density under Gaussian-beam excitation

```python
x_beam, y_beam = width / 2, width / 2
w_beam = 1e-6  # m

n = solve_exciton_diffusion(strain, width, width, x_beam=x_beam, y_beam=y_beam, w_beam=w_beam)
```

## Plot the exciton density map

```python
ext = [0, width * 1e6, 0, width * 1e6]  # μm

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

vmax = np.abs(strain).max()
im0 = axes[0].imshow(strain*100, cmap='coolwarm', origin='lower', extent=ext,
                      aspect='equal')
axes[0].set_title('Strain (%)')
axes[0].set_xlabel('μm')
axes[0].set_ylabel('μm')
plt.colorbar(im0, ax=axes[0], fraction=0.046)

im1 = axes[1].imshow(n, cmap='coolwarm', origin='lower', extent=ext, aspect='equal')
axes[1].set_title('Exciton density N(x, y)')
axes[1].set_xlabel('μm')
axes[1].plot(x_beam * 1e6, y_beam * 1e6, 'w+', markersize=10, markeredgewidth=2)
plt.colorbar(im1, ax=axes[1], fraction=0.046)

plt.tight_layout()
plt.show()
```
