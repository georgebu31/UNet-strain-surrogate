# FDM solver demo
Runs the Kirchhoff-Love plate solver on a single example substrate geometry
and plots the resulting support map and strain field.

```python
import numpy as np
import matplotlib.pyplot as plt

from fdm_solver import get_w_strain, draw_circle
```

## Build an example substrate (single circular pillar) and solve for strain

```python
NPZ_FILENAME   = "swg_256.npz"
NPZ_ARRAY_NAME = 'array'
width  = 12e-6
height = 12e-6

support_data = np.load(NPZ_FILENAME)
if NPZ_ARRAY_NAME in support_data:
    support_map = support_data[NPZ_ARRAY_NAME]
else:
    support_map = support_data[support_data.files[0]]

support_map = np.where(support_map > 0.5, 1, 0)

strain, X, Y = get_w_strain(support_map, width)
```

## Plot support map and strain field

```python
ext = [0, width * 1e6, 0, width * 1e6]  # μm

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].imshow(support_map, cmap='gray_r', origin='lower', extent=ext, aspect='equal')
axes[0].set_title('Support map')
axes[0].set_xlabel('μm')
axes[0].set_ylabel('μm')

vmax = np.abs(strain).max()
im = axes[1].imshow(strain*100, cmap='coolwarm', origin='lower', extent=ext,
                     aspect='equal')
axes[1].set_title('Strain (%)')
axes[1].set_xlabel('μm')
plt.colorbar(im, ax=axes[1], fraction=0.046)

plt.tight_layout()
plt.show()
```
